Import Libraries

```python
import numpy as np
```

Define Global Model

```python
class LinearModel:
    def __init__(self):
        self.w = np.random.randn()
        self.b = np.random.randn()
```

Client-Side Training

```python
def train_local_model(model, X, y, lr=0.01, epochs=50):
    w, b = model.w, model.b

    for _ in range(epochs):
        y_pred = w * X + b
        dw = np.mean((y_pred - y) * X)
        db = np.mean(y_pred - y)

        w -= lr * dw
        b -= lr * db

    return w, b
```

Federated Averaging (Server Aggregation)

```python
def federated_averaging(client_weights):
    avg_w = np.mean([w for w, b in client_weights])
    avg_b = np.mean([b for w, b in client_weights])
    return avg_w, avg_b
```

Simulate Federated Learning

```python
def federated_learning(num_clients=5, rounds=5):
    # Create synthetic local data for each client
    clients_data = []
    for i in range(num_clients):
        X = np.random.rand(50) * 10
        y = 2 * X + 3 + np.random.randn(50)  # true model
        clients_data.append((X, y))

    # Initialize global model
    global_model = LinearModel()

    print("Initial Global Model:")
    print(f"w = {global_model.w:.4f}, b = {global_model.b:.4f}\n")

    for round_num in range(rounds):
        client_updates = []

        for X, y in clients_data:
            local_model = LinearModel()
            local_model.w = global_model.w
            local_model.b = global_model.b

            updated_w, updated_b = train_local_model(local_model, X, y)
            client_updates.append((updated_w, updated_b))

        # Server aggregation
        global_model.w, global_model.b = federated_averaging(client_updates)

        print(f"Round {round_num+1}: Global w = {global_model.w:.4f}, b = {global_model.b:.4f}")

    return global_model
```

Run the Simulation

```python
final_model = federated_learning()
```
